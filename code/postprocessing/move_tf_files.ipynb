## Check if parallelization code gives the same results as serial

```python
glob(f"derivatives/preprocessed/TF_outputs/main/seed_fixed*.mat")
```

```python
m = "TF"
for s in ["3000001", "3000002", "3000025", "3000131"]:
    for c in [#"resp_s_i_0",
              "resp_s_i_1", "resp_s_c_1", "resp_ns_i_0", "resp_ns_i_1", "resp_ns_c_1"]:
        run_1_files = sorted(glob(f"derivatives/preprocessed/TF_outputs/main/resp/{m}/sub*{s}*{m}*{c}*.mat"))
        run_2_files = sorted(glob(f"derivatives/preprocessed/TF_outputs/main/resp/sub*{s}*{m}*{c}*.mat"))
        
        print(c, f"\n\nBefore :{len(run_1_files)}\n",
        f"After :{len(run_2_files)}\n")
    
        file_1 = run_1_files[0]
        print(file_1)
        f_1 = h5py.File(file_1)
        print(f_1.keys())
        
        key = list(f_1.keys())[-1]
        
        file_2 = run_2_files[0]
        print(file_2)
        f_2 = h5py.File(file_2)
        print(f_2.keys())
        print(
            np.mean(f_1[key][:]), np.mean(f_2[key][:])
        )
        print(
            np.mean(f_1[key][:]) == np.mean(f_2[key][:])
        )
```
